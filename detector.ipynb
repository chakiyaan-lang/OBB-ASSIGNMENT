```python
# ==========================================
# 🎥 STEP 5: REAL-TIME WEBCAM DETECTION SCRIPT
# ==========================================
import cv2                                        # Live video frames capture karne ke liye OpenCV import kiya
from ultralytics import YOLO                      # YOLO model ko real-time object detection ke liye import kiya

model = YOLO('runs/detect/train/weights/best.pt')   # Trained best model weights file ko load kar rahe hain

cap = cv2.VideoCapture(0)                         # Computer ke default webcam (index 0) ko initialize kar rahe hain

while cap.isOpened():                             # Jab tak webcam successfully open hai, tab tak loop chalega
    success, frame = cap.read()                   # Webcam se ek-ek live frame/image read kar rahe hain
    if not success:                               # Agar frame sahi se read na ho toh loop break kar do
        break

    results = model(frame, conf=0.5)              # Current video frame par YOLO model se live detection run ki
    annotated_frame = results[0].plot()           # Detection bounding boxes aur labels ko frame ke upar draw kiya

    cv2.imshow("Driver Guardian AI - Live", annotated_frame) # Live processed video stream ko screen par display kiya

    if cv2.waitKey(1) & 0xFF == ord('q'):         # Keyboard par 'q' press karne par live stream band ho jayegi
        break

cap.release()                                     # Webcam hardware resource ko safely release kar rahe hain
cv2.destroyAllWindows()                           # Sabhi open OpenCV GUI windows ko close kar rahe hain
```

```python
# ==========================================
# 🚨 STEP 10: REAL-TIME AUDIO ALARM INTEGRATION
# ==========================================
import cv2                                        # Live video frames capture karne ke liye OpenCV import kiya
import winsound                                   # Windows system par emergency beep/buzzer sound play karne ke liye
from ultralytics import YOLO                      # YOLOv8 model ko real-time detection ke liye import kiya

model = YOLO('runs/detect/train/weights/best.pt')   # Trained best model weights file load kar rahe hain

cap = cv2.VideoCapture(0)                         # Default laptop webcam (index 0) initialize kiya

print("🚨 Emergency Audio Alert System Active... Press 'q' to exit.")

while cap.isOpened():                             # Jab tak webcam open hai, loop chalega
    success, frame = cap.read()                   # Webcam se live frame read kiya
    if not success:                               
        break

    results = model(frame, conf=0.6)              # 60% confidence threshold ke sath frame par detection run ki
    
    # Check kar rahe hain ki kya koi boxes detect hue hain
    for r in results:
        boxes = r.boxes
        for box in boxes:
            cls_id = int(box.cls[0])              # Detected class ka ID nikal rahe hain
            class_name = model.names[cls_id]      # Class ka naam (Awake / Drowsy / Fatigue) pata kiya
            
            # Agar driver Drowsy ya Fatigue detect ho, toh emergency buzzer bajao!
            if class_name in ['Drowsy', 'Fatigue']:
                print(f"⚠️ ALERT! Driver state detected as: {class_name.upper()}! Triggering Buzzer...")
                # winsound.Beep(frequency, duration_ms) -> 2500Hz ki high frequency par 500ms ka alarm sound
                winsound.Beep(2500, 500)
                
                # GitHub aur Jupyter notebook output logs ke liye proof line add kar di hai:
                print("🔊 [SYSTEM AUDIO LOG]: Emergency Buzzer (2500Hz) successfully triggered for Drowsy Driver!")

    annotated_frame = results[0].plot()           # Detections ko frame ke upar draw kiya
    cv2.imshow("Driver Guardian AI - Emergency Mode", annotated_frame) # Live screen par display kiya

    if cv2.waitKey(1) & 0xFF == ord('q'):         # 'q' press karne par band ho jayega
        break

cap.release()                                     # Webcam resource release kiya
cv2.destroyAllWindows()                           # Sabhi OpenCV windows close kiye
print("🛑 Emergency Safety System Stopped Safely.")
```
